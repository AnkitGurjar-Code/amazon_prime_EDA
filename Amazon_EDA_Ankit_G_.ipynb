## 4. ***Data Wrangling***

### Data Wrangling Code

````python
# Check Unique Values for each variable and handle missing values for 'age_certification'
# The .unique() method is used to inspect current unique values before filling.
# Using inplace=True to modify the DataFrame directly.
titles['age_certification'].fillna('not rated', inplace=True)

# Filling null values with the mean of 'imdb_score' column
# This is a common imputation technique for numerical columns.
titles['imdb_score'].fillna(titles['imdb_score'].mean(), inplace=True)

# Dropping duplicate rows from both title and credit dataframes
# This ensures that each row is unique, which is good practice for data cleanliness.
titles.drop_duplicates(inplace=True)
credits.drop_duplicates(inplace=True)
````

````python
# Write your code to make your dataset analysis ready.
# Merge datasets
merged_df = pd.merge(titles, credits, on='id', how='left')
````

### What all manipulations have you done and insights you found?

In the data wrangling step, both datasets were merged using the 'id' column to combine all the relevant information into a single dataframe. After merging, the dataset was checked using basic functions like head() and info() to understand its structure.

Duplicate records were removed to maintain data consistency. Missing values were handled by filling them with 'Unknown' so that further analysis does not face any issues. The 'release_year' column was converted into numeric format to ensure proper analysis. Additionally, the 'seasons' column was dropped as it was not required for the analysis.

From this process, it was understood that the dataset had missing values and some unnecessary columns, which were cleaned to make the data ready for visualization and analysis.

#### Chart - 5 Trend of Content Releases

````python
# Chart - 5 visualization code

data = merged_df.copy()

# 🧹 Clean and prepare data
data['release_year'] = pd.to_numeric(data['release_year'], errors='coerce')
data = data.dropna(subset=['release_year', 'type'])
data = data[data['release_year'] >= 1980]  # Filter relevant years

# ✅ Standardize 'type' column
data['type'] = data['type'].str.strip().str.title()  # Fix casing
data['type'] = data['type'].replace({
    'Tv Show': 'TV Show',
    'Show': 'TV Show',
    'Movie ': 'Movie'
})

# 📊 Group by release year and content type
trend_data = (
    data.groupby(['release_year', 'type'])
    .size()
    .reset_index(name='count')
)

# 🎨 Set style
sns.set_theme(style="whitegrid")

plt.figure(figsize=(12, 7))

# 📈 Line Plot
sns.lineplot(
    data=trend_data,
    x='release_year',
    y='count',
    hue='type',
    marker='o',
    linewidth=2.5,
    palette={'Movie': '#1f77b4', 'TV Show': '#ff7f0e'}
)

# 🏷 Titles and labels
plt.title("Trend of Content Releases on Amazon Prime (Movies vs TV Shows)",
          fontsize=18, fontweight="bold", color="#2E4053", pad=15)
plt.xlabel("Release Year", fontsize=14, color="#1a5276")
plt.ylabel("Number of Titles Released", fontsize=14, color="#1a5276")
plt.xticks(rotation=45, fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title="Content Type", title_fontsize=13, fontsize=12, loc="upper left")

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A line chart was chosen because it clearly visualizes how movie and TV show releases have changed over time.

##### 2. What is/are the insight(s) found from the chart?

Both Movies and TV Shows show growth over the years, but Movies remain more dominant. TV Show production has also increased steadily in recent years.

#### Chart - 6 Average IMDb Score by Top 10 Genres

````python
# Chart - 6 visualization code
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = merged_df.copy()

# 🧹 Clean and standardize 'type' column
data['type'] = data['type'].astype(str).str.strip().str.title()

# Replace all common variants of show/movie
data['type'] = data['type'].replace({
    'Tv Show': 'TV Show',
    'Show': 'TV Show',
    'Tv-Show': 'TV Show',
    'Tvshow': 'TV Show',
    'Movies': 'Movie',
    'Movie ': 'Movie',
    'Film': 'Movie'
})

# Drop invalid/unknown values
valid_types = ['Movie', 'TV Show']
data = data[data['type'].isin(valid_types)]

# Sanity check
print("✅ Unique type values after cleaning:", data['type'].unique())

# Detect genre column
genre_col = "genres"

# Clean genre data
data[genre_col] = data[genre_col].fillna("Unknown")
data[genre_col] = data[genre_col].apply(lambda x: x.split(",")[0].strip() if isinstance(x, str) else x)

# Filter valid IMDb scores
data = data.dropna(subset=[genre_col, "imdb_score"])
data = data[(data["imdb_score"] > 0) & (data["imdb_score"] <= 10)]

# Top 10 Genres
top_genres = data[genre_col].value_counts().head(10).index
filtered_data = data[data[genre_col].isin(top_genres)]

# Group by Genre and Type
grouped_data = (
    filtered_data.groupby([genre_col, "type"], as_index=False)["imdb_score"]
    .mean()
    .sort_values(by="imdb_score", ascending=False)
)






# Plot
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 8))

bar_plot = sns.barplot(
    data=grouped_data,
    x=genre_col,
    y="imdb_score",
    hue="type",
    dodge=True
)



# Title & Labels
plt.title("Average IMDb Score by Top 10 Genres (Movies vs TV Shows)",
          fontsize=18, fontweight="bold", color="#2E4053", pad=20)
plt.xlabel("Genre", fontsize=14, color="#1A5276")
plt.ylabel("Average IMDb Score", fontsize=14, color="#1A5276")
plt.xticks(rotation=30, fontsize=12)
plt.legend(title="Content Type", title_fontsize=13, fontsize=12)
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A bar chart is effective for comparing average IMDb ratings across different genres and identifying the highest-rated categories.

##### 2. What is/are the insight(s) found from the chart?

Some genres consistently receive higher IMDb ratings than others, indicating stronger audience appreciation and better content quality.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. Amazon Prime can focus on producing or acquiring content from highly rated genres to improve viewer satisfaction. However, focusing only on highly rated genres may reduce content diversity.

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
# Calculate the average IMDb score for each genre
genre_avg_imdb_score = merged_df.groupby('genres')['imdb_score'].mean()

# Get the top 15 genres based on their average IMDb score
top_15_genres = genre_avg_imdb_score.nlargest(15).index
# Filter the merged_df to include only the top 15 genres
filtered_df = merged_df[merged_df['genres'].isin(top_15_genres)]

# Create the pivot table using the filtered DataFrame
genre_pivot = filtered_df.pivot_table(index='genres', columns='type', values='imdb_score', aggfunc='mean')

plt.figure(figsize=(12, 8))
sns.heatmap(genre_pivot, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Top 15 Genres by Average IMDb Score Across Content Types')
plt.xlabel('Content Type')
plt.ylabel('Genre')
plt.show()
````

##### 1. Why did you pick the specific chart?

A heatmap is useful for identifying relationships and correlation strengths among multiple variables simultaneously.

##### 2. What is/are the insight(s) found from the chart?

Some variables show positive correlations while others show weak or negligible relationships. This helps identify factors that may influence content performance.
